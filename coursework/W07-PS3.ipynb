We've already introduced you to **lists** and **tuples**, but now we're going to show you another variable type which can store multiple values: **dictionaries**

Unlike lists and tuples, the data in dictionaries is stored in pairs consisting of a **key** and **value**.

A cricket fan might want to store the names of captains of IPL teams using a dictionary where:

- Team name would be the **key**
- Captain's name would be the **value**

You can think of a dictionary as a bit like a two column table:

| key | value |
| :-- | :-: |
| CSK | MS Dhoni |
| MI  | Rohit Sharma |
| RCB | Virat Kohli  |
| SRH | David Warner |
| KXIP| Lokesh Rahul |

Here are a few rules about dictionaries:

- Flexible data type - Both the key and value are pretty flexible in the data type which is stored in them. For example, you could use a string, integer, float or even a tuple as a key or a value.
- No duplicate keys - You cannot have the same key entered twice - this is because you look up data using the key, so each key needs to unambiguously refer to one value.
- You can have duplicate values (eg. you could enter the same captain for two teams - although they might be a bit busy)



#### 1. Declaring a dictionary

The syntax for storing data in a dictionary is:

- use curly brackets {}
- separate the key and value with a colon :
- separate the pairs with commas ,

The code below stores the data on IPL captains in a dictionary variable named **IPL_captains**.

```python
IPL_captains = {'CSK' : 'MS Dhoni', 'MI' : 'Rohit Sharma',
                'RCB' : 'Virat Kohli', 'SRH' : 'David Warner', 
                'KXIP' : 'Lokesh Rahul'}

print(IPL_captains)
```

#### 2. Accessing the dictionary - part 1

The advantage of having a the keys in the dictionary, is that you can look up values with them. So if you wanted to know the captain of the Mumbai Indians (MI), you don't have to loop through the whole dictionary until you find a match, you just enter the key and the value is returned, like this:

```python
IPL_captains['MI']
```

Now it's your turn. Look up the captain of the Sunrisers Hyderabad (SRH).

```python
# look up SRH captain
IPL_captains['SRH']
```

#### 3. Adding to and updating a dictionary

Adding new key:value pairs to a dictionary, and updating an existing value are both straightforward, and are done in the same way.

You could add the Delhi Capitals (DC) captain Shreyas Iyer to the dictionary with the following command:

```python
IPL_captains['DC'] = 'Shreyas Iyer'

print(IPL_captains)
```

If David Warner's groin injury gets worse he might be replaced as skipper by Kane Williamson. The dictionary could be updated like so:

```python
IPL_captains['SRH'] = 'Kane Williamson'

print(IPL_captains)
```

You can see that the commands to add to and update the dictionary are exactly the same. What happens depends on whether that key is present in the dictionary.

- If the key is present, the associated value is updated.
- If the key is not present, the new key:value pair are added to the end of the dictionary.

#### 4. Accessing the dictionary - part 2

With previous collections of items, like lists and tuples, you've been able to access them using a numerical index.

Just as a reminder, look at the list of sports in the box below. You can access the items in the list by numerical index like so:

```python
sport_list = ['hockey', 'cricket', 'fencing']

sport_list[2]
```

```python
IPL_captains[1]
```

Look at the error message you got. Next try to access a non-existant key - perhaps my fantasy team the 'London Bridge Lancers (LBL)'.

- Use 'LBL' as the key to lookup the value in IPL_captains

```python
# lookup key 'LBL'
IPL_captains['LBL']
```

Now compare the two error messages - they should be the same: KeyError

This means the key was not found in the dictionary. So when you entered `IPL_captains[1]` it was actually looking for a key that was `1`, rather than item number 1. Because you can use integers as keys in a dictionary, Python will assume you are looking for a key if you enter an integer between the square brackets.

Which raises the question of how do you loop through a dictionary? You can do this using the `items()` function.

```python
for item in IPL_captains.items():
    print(item)
```

That allows you to loop through the key:value pairs. But there are also the `keys()` and `values()` functions which access them separately.

In the box below, write 2 for loops:

- loop through the keys of IPL_captains and print them
- loop through the values of IPL_captains and print the length of the captain's names

```python
# loop through keys and print
for key in IPL_captains.keys():
    print(key)

# loop through values and print length
for value in IPL_captains.values():
    print(len(value))

```
