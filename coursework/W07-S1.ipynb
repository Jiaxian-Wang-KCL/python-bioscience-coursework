Each UK railway station has a unique three letter code. Here are the codes for some london stations:

| code | station         |
| :--  | :-:              |
| WAT  | Waterloo         |
| KGX  | Kings Cross     |
| VIC  | Victoria         |
| VXH  | Vauxhall         |
| CLJ  | Clapham Junction |

1. Create a dictionary of the station codes and names.

```python
London_Station = {"WAT":"Waterloo",
                  "KGX":"Kings Cross",
                  "VIC":"Victoria",
                  "VXH":"Vauxhall",
                  "CLJ":"Clapham Junction"}
```

2. Write code to ask the user to enter a code, then print the corresponding station name.

```python
ask_code = input("Enter a code")
print(London_Station[ask_code])
```

```python
London_Station["MAY"] = "Marylebone"
```

4. Print all the keys from the dictionary.

5. Now write a function that:
