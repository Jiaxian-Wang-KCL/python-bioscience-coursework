2. Generate two lists of numbers as follows:
- x: going from 0 to 9 in increments of 1
- y: going from 10 to 100 in increments of 10

Plot the two lists using pyplot.

```python
### SOLUTION ###

# load library
import matplotlib.pyplot as plt

# initiate lists
x = []
y = []

# loop appending to lists
for i in range(10):
    x.append(i)
    
for i in range(10, 110, 10):
    y.append(i)
    
# plot lists
plt.plot(x, y)
plt.show()
```

3. Generate a list starting with 1 and containing 10 values, where each subsequent value is double the previous value.

*Hint: you can either store the current value in a separate variable or lookup the previous value in the list.*

Print the following values:
- whole list
- first item
- last item
- third item
- item in position 3

```python
### SOLUTION ###

# initiate list
dublist = [1]

# generate list
for i in range(1, 10):
    dublist.append(dublist[i-1] * 2)
    
# whole list
print(dublist)

# first item
print(dublist[0])

# last item
print(dublist[-1])

# third item
print(dublist[2])

# item 3
print(dublist[3])
```

```python
### ALTERNATIVE SOLUTION ###

# initiate list
dublist = []
x = 1

# generate list
for i in range(10):
    dublist.append(x)
    x = 2 * x
    
# whole list
print(dublist)

# first item
print(dublist[0])

# last item
print(dublist[-1])

# third item
print(dublist[2])

# item 3
print(dublist[3])
```

5. Plot a graph of $y = x^3$

*Hint: you will have to generate lists for x and y first.*

```python
### SOLUTION ###

# load library
import matplotlib.pyplot as plt

# initiate lists
x = []
y = []

# generate lists
for i in range(100):
    x.append(i)
    y.append(i**3)
    
# plot lists
plt.plot(x, y)
plt.show()
```
