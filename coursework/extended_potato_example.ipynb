```python
# open and read the message
file = open("secretmessage.txt", "r")
secrettext = file.read()
file.close()

print(secrettext)
length_secret = len(secrettext)
print(length_secret)

potatoflag = False
found_first_space = False
found_second_space = False

# find where "potato" starts
for i in range(0, length_secret - 5):
    newpotatostr = secrettext[i:i+6]
    if newpotatostr == "potato":
        position_potato = i
        potatoflag = True

if potatoflag == True:

    # find the first space (directly before "potato")
    for i in range(0, position_potato):
        newindex = position_potato - 1 - i
        if secrettext[newindex] == " " and found_first_space == False:
            first_space_index = newindex
            found_first_space = True

        # find the second space (the one before that)
        elif secrettext[newindex] == " " and found_first_space == True and found_second_space == False:
            second_space_index = newindex
            found_second_space = True

    # once both spaces are found, slice the string between them
    if found_first_space == True and found_second_space == True:
        newstring = secrettext[second_space_index + 1 : first_space_index]
        print(newstring)
```
