That is a good general function to have, but in this case it would probably be easier to use a function where we just have to specify the PAM start position, and it returns the gRNA sequence.

```python
def extract_grna(sequence, pam_start, inc_pam=False):
    # gRNA always starts 20 nt BEFORE the PAM
    start = pam_start - 20
    
    # length depends on whether we include the PAM
    if inc_pam:
        length = 23     # 20 nt spacer + 3 nt PAM
    else:
        length = 20     # spacer only
    
    return sequence[start:start + length]

file = open("dnaseq2.txt", "r")
dnaseq = file.read()
file.close()

pam_positions = find_pam(0,0)

print("PAM positions:", pam_positions)

if pam_positions:
    pam_start = pam_positions[0]

    # extract with PAM included
    seq_with_pam = extract_grna(dnaseq, pam_start, inc_pam=True)
    print("gRNA + PAM (23 nt):", seq_with_pam)

    # extract without PAM
    seq_without_pam = extract_grna(dnaseq, pam_start, inc_pam=False)
    print("gRNA only (20 nt):", seq_without_pam)

else:
    print("No PAM found")
```

### 7. Reverse complement using a dictionary I

Now you will move onto a function to reverse complement a DNA sequence. Take a look at the code below. It reads through the a DNA sequence in reverse order from the last nucleotide to the first nucleotide and replaces each nucleotide with the complementary nucleotide. Check that you understand what each section of code does.

```python
# read file contents into dnaseq
filecontents = open('dnaseq.txt', 'r')
sequence = filecontents.read()
filecontents.close()

# remove \n
sequence = sequence.strip()

# calculate sequence length
seq_len = len(sequence)

# empty variable for reverse complement
rc_seq = ""

# loop through seq in reverse
for i in range(seq_len - 1, -1, -1):
    if sequence[i] == "a":
        rc_seq = rc_seq + "t"
    elif sequence[i] == "t":
        rc_seq = rc_seq + "a"
    elif sequence[i] == "g":
        rc_seq = rc_seq + "c"
    elif sequence[i] == "c":
        rc_seq = rc_seq + "g"

# print sequence in both orientations
print(sequence)
print(rc_seq)
```

You can see that the combination of if & elif statements to identify the complementary nucleotides is a bit cumbersome. This is a good example of where a **dictionary** can make your code shorter and clearer. The four nucleotides and their complementary nucleotides can be stored as pairs in a dictionary, and this can be used to identify the complementary nucleotides in a single line of code.

### 10. GC Content function

Your final function today will be to calculate the GC content of a sequence. Take a look at the code below and check you understand how it works.

Copy the code into the empty cell below and modify it to define a function named **gc_content** which takes a single argument **sequence**. Your function should return the same value as the code below.

```python
def gc_content(sequence):
    # Remove whitespace / newlines just in case
    sequence = sequence.strip()

    seq_len = len(sequence)
    gc_count = 0

    # Count G and C nucleotides
    for i in range(seq_len):
        if sequence[i] == "g" or sequence[i] == "c":
            gc_count += 1

    return gc_count / seq_len
filecontents = open('dnaseq.txt', 'r')
dnaseq = filecontents.read()
filecontents.close()

print(gc_content(dnaseq))
```

### 11. Now put it all together

Now that you've got all the functions working separately, you should put them together to create a gRNA search program. A suggested program flow is:

- define all the functions you will need in the rest of the program (just copy and paste from the cells above)
- read in the contents of file **dnaseq2.txt**
- set variables to define the region 300 - 600 nt to search within
- search the forward strand of this region for PAMs and store the positions in a list
- use this list to extract the gRNA sequences
- reverse complement the whole dna sequence and store the reverse strand
- calculate the region of the reverse strand to search within
- search for PAMs in this region of the reverse strand and store in a new list
- using this list to extract the gRNA sequences from the reverse strand
- combine the two lists of gRNA sequences into one list
- loop through the list of gRNA sequences and check the GC content - if they are between 40 - 60% GC content, store them in a final list
- return the list of passed gRNA sequences

```python
# ---------------------------
# 1. Define all required functions
# ---------------------------

# Complement dictionary
dna_dict = {
    "a": "t",
    "t": "a",
    "g": "c",
    "c": "g"
}

# Reverse complement function
def rev_comp(sequence):
    sequence = sequence.strip()
    seq_len = len(sequence)
    rc_seq = ""
    for i in range(seq_len - 1, -1, -1):
        rc_seq = rc_seq + dna_dict[sequence[i]]
    return rc_seq

# PAM search function
def find_pam(sequence, start, stop):
    pam_positions = []
    
    # avoid going out of range
    if stop > len(sequence) - 3:
        stop = len(sequence) - 3
    if start < 20:
        start = 20
    
    for i in range(start, stop):
        # PAM = NGG, so check the last two nucleotides
        if sequence[i+1] == "g" and sequence[i+2] == "g":
            pam_positions.append(i)
    
    return pam_positions

# extract gRNA (20nt) or gRNA+PAM (23nt)
def extract_grna(sequence, pam_start, inc_pam=False):
    start = pam_start - 20
    length = 23 if inc_pam else 20
    return sequence[start:start + length]

# GC content function
def gc_content(sequence):
    sequence = sequence.strip()
    seq_len = len(sequence)
    gc_count = 0
    for i in range(seq_len):
        if sequence[i] == "g" or sequence[i] == "c":
            gc_count += 1
    return gc_count / seq_len


# ---------------------------
# 2. Read DNA sequence from file
# ---------------------------

filecontents = open("dnaseq2.txt", "r")
dnaseq = filecontents.read().strip()
filecontents.close()


# ---------------------------
# 3. Define region 300–600 to search
# ---------------------------

search_start = 300
search_end   = 600


# ---------------------------
# 4. Forward strand PAM search
# ---------------------------

forward_pams = find_pam(dnaseq, search_start, search_end)

# Extract forward gRNAs
forward_grnas = []
for pam in forward_pams:
    grna = extract_grna(dnaseq, pam, inc_pam=True)   # include PAM
    forward_grnas.append(grna)


# ---------------------------
# 5. Reverse complement whole sequence
# ---------------------------

rev_dnaseq = rev_comp(dnaseq)

# region on reverse strand corresponds to reversed coordinates
rev_search_start = len(dnaseq) - search_end
rev_search_end   = len(dnaseq) - search_start


# ---------------------------
# 6. Reverse strand PAM search
# ---------------------------

reverse_pams = find_pam(rev_dnaseq, rev_search_start, rev_search_end)

# Extract reverse-strand gRNAs
reverse_grnas = []
for pam in reverse_pams:
    grna = extract_grna(rev_dnaseq, pam, inc_pam=True)
    reverse_grnas.append(grna)


# ---------------------------
# 7. Combine all gRNAs
# ---------------------------

all_grnas = forward_grnas + reverse_grnas


# ---------------------------
# 8. Filter by GC content (40–60%)
# ---------------------------

passed = []
for grna in all_grnas:
    gc = gc_content(grna)
    if 0.40 <= gc <= 0.60:
        passed.append(grna)


# ---------------------------
# 9. Return (print) the final list of valid gRNAs
# ---------------------------

print("Final gRNAs (40–60% GC):")
print(passed)
print("\nNumber of valid gRNAs:", len(passed))
```
