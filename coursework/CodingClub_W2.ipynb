   - Append 10

7. Create an empty list **evens**, loop through numbers 1–20, and append only the even ones.

````python
values = [3,9,1,4]
smallest_value = min(values)
print(smallest_value)
for i in range(len(values)):
    if values[i] == smallest_value:
        s_index = i 
values.pop(s_index)
values.append(10)
print(values)

evens = []
for i in range(1,21):
    if i%2 == 0:
        evens.append(i)
print(evens)
````
