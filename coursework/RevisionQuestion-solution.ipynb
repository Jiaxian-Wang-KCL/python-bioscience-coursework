## Restriction enzyme finder

Restriction enzymes bind to specific DNA sequences and cut the DNA strand. The table below shows the recognition sequences for some common restriction enzymes with the cut site indicated by the slash (/). For example, BamHI recognises the sequence `GGATCC` and cuts between the first two nucleotides of that sequence.

**Table 1: Restriction Enzymes Recognition Sequences**

| Sequence | Enzyme|
|:---------|-------|
| G/GATCC  | BamHI |
| G/AATTC  | EcoRI |
| GAT/ATC  | EcoRV |
| G/CTAGC  | NheI  |
| T/CTAGA  | XbaI  |

3. Write a function to identify the cuts sites for any of the enzymes in table 1. The user should provide the function with a DNA sequence and the name of the enzyme. The function should return a list containing the position of the cut sites.

````python
def findCutSite(sequence, re_name):
    # finds restriction enzyme sites in the DNA sequence provided
    # inputs: DNA sequence, restriction enzyme name
    # output: list of cut sites

    # define RE site dictionary
    re_dict = {'BamHI':'G/GATCC', 'EcoRI':'G/AATTC', 'EcoRV':'GAT/ATC', 'NheI':'G/CTAGC', 'XbaI':'T/CTAGA'}

    # empty list to store cut sites
    cut_sites = []
    
    # process recognition site
    re_site = re_dict[re_name]

    # empty string for processed site
    re_site_proc = ''

    # loop through re site find cut position and remove /
    for i in range(len(re_site)):
        if re_site[i] == '/':
            cut_position = i - 1
        else:
            re_site_proc += re_site[i]

    # search sequence for recognition site
    for i in range(len(sequence) - len(re_site_proc) + 1):
        if sequence[i:i + len(re_site_proc)] == re_site_proc:
            # calculate cut position
            re_cut_position = i + cut_position
            cut_sites.append(re_cut_position)

    # return list of cut sites
    return cut_sites

findCutSite(seq, 'EcoRV')
````

#### Restriction enzymes with amiguity codes

Some restriction enzymes have ambiguous recognition sequences where more than one type of nucleotide can be accommodated at certain positions in the recognition sequence.

**Table 2: Restriction Enzymes Recognition Sequences**

| Sequence     | Enzyme|
|:-------------|-------|
| G/GYRCC      | BanI  |
| GCCNNNN/NGGC | BglI  |
| W/CCGGW      | BsaWI |

**Table 3: IUPAC Ambiguity Codes for Nucleotides**

| Code | Bases      |
|:----:|------------|
| A    | A          |
| C    | C          |
| G    | G          |
| T    | T          |
| W    | A or T     |
| S    | C or G     |
| R    | A or G     |
| Y    | C or T     |
| N    | A C G or T |

4. Write a function to identify the cut sites for restriction enzymes with ambiguous recognition sequences. Your function should take the recognition sequence exactly as it is written in Table 2 and the name of a sequence file to search. For each cut site found, your function should print the position of the cut site, and the 5 nucleotides either side of each cut site with a space representing where the DNA strand has been cut.

````python
def ambifinder(recog_seq, seqfile):
    # searches for ambiguous recognition sequences in the given sequence file

    # read sequence from file
    file = open(seqfile)
    sequence = file.read().strip()
    file.close()

    # store ambiguity codes
    ambi_dict = {'N':['A', 'C', 'G', 'T'], 'Y':['C', 'T'], 'R':['A', 'G'], 'W':['A','T'], 'S':['C', 'G'], 'A':['A'], 'C':['C'], 'G':['G'], 'T':['T']}

    # process recognition sequences
    recog_seq_proc = ''
    for i in range(len(recog_seq)):
        if recog_seq[i] == '/':
            cut_position = i - 1
        else:
            recog_seq_proc += recog_seq[i]

    # ambiguous search sequence
    # loop through main sequence
    for i in range(len(sequence) - len(recog_seq_proc) + 1):

        # flag variable to count matches
        match_count = 0

        # loop through recognition sequence
        for j in range(len(recog_seq_proc)):

            # ambiguous matching
            ambi_code = recog_seq_proc[j]
            ambi_list = ambi_dict[ambi_code]
            if sequence[i + j] in ambi_list:
                match_count += 1
                
        # check whether whole recognition sequence matched
        if match_count == len(recog_seq_proc):
            # identified recognition sequence and calc cut site
            recog_cut_site = i + cut_position    

            # print required information
            print('cut site at:', recog_cut_site)
            print(sequence[recog_cut_site - 4:recog_cut_site + 1], sequence[recog_cut_site + 1:recog_cut_site + 6])

# call the function and test with different recognition sequences
ambifinder('G/GYRCC', 'seq1.txt')
ambifinder('GCCNNNN/NGGC', 'seq1.txt')
ambifinder('W/CCGGW', 'seq1.txt')
````
