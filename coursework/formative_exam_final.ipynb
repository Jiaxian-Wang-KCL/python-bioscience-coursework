1a. Discuss the differences between for loops and while loops, and for which types of problems you might use each of them. 15

Write your answer here:
A for loop is used when the number of iterations is known in advance. It repeats a block of code for each element in a sequence (such as a list, string, or range). For example, a for loop can be used to print all numbers from 1 to 10 or to process every item in a dataset. It is most useful for problems where the amount of data or the number of repetitions is fixed and predictable.

A while loop, on the other hand, is used when the number of iterations is not known beforehand. It continues to execute a block of code as long as a specified condition remains true. For example, a while loop can be used to keep asking a user for input until they type a valid answer or to simulate a biological process until equilibrium is reached. It is best for problems that depend on a changing condition rather than a fixed number of steps.

2 a. The sequence in DNAsequence.txt comes from yeast which reproduces asexually, meaning that changes in DNA occur only through mutation. For each reproduction cycle, each base has a 1 in 10,000 chance of being mutated to a different base. Simulate the random mutation process for 1000 reproduction cycles. When the base has been mutated, replace the letter by "M". 25

```python
import random
# open the file with read mode
file = open('DNAsequence_formative.txt','r')
#read the file
DNAseq = file.read().strip()
#close the file
file.close()
#fine the length of DNA sequence
length_DNAseq = len(DNAseq)
# Convert the sequence to a list, as strings are immutable
dna_list = []
for base in DNAseq:
    dna_list.append(base)
#Simulate 1000 breeding cycles 
cycles = 1000
mutation_rate = 1/10000
for cycle in range(cycles):
    # For each base in the sequence
    i = 0
    while i < len(dna_list):
        # Only examine bases that have not yet mutated (not 'M')
        if dna_list[i] != 'M':
            # Generate a random number between 0 and 1 inclusive
            random_number = random.random()
            # If the random number is less than the mutation rate, a mutation occurs.
            if random_number < mutation_rate:
                dna_list[i] = 'M'  # Marked as mutated
        i += 1
# Display results
result_sequence = ''
for i in dna_list:
    result_sequence += i
print(result_sequence)
```

2 c. Use your mutation simulation, and check whether after 1000 reproduction cycles the sequence in DNAsubsequence_formative.txt is still present within DNAsequence, or whether it has been mutated. 10

```python
import random

# Read main sequence file
file1 = open('DNAsequence_formative.txt', 'r')
main_sequence = file1.read().strip()
file1.close()

# Read subsequence file
file2 = open('DNAsubsequence_formative.txt', 'r')
subsequence = file2.read().strip()
file2.close()

# Find the position of subsequence in main sequence
start_index = main_sequence.find(subsequence)

# Convert main sequence to list for mutation simulation
dna_list = []
for base in main_sequence:
    dna_list.append(base)

# Simulate 1000 reproduction cycles
cycles = 1000
mutation_rate = 1/10000

for cycle in range(cycles):
    i = 0
    while i < len(dna_list):
        # Only check bases that haven't mutated yet
        if dna_list[i] != 'M':
            # Generate random number and check if mutation occurs
            random_number = random.random()
            if random_number < mutation_rate:
                dna_list[i] = 'M'  # Mark as mutated
        i += 1

# Convert mutated sequence back to string
mutated_sequence = ''
for i in dna_list:
    mutated_sequence += i
print(mutated_sequence)

# Extract the subsequence region from mutated sequence
subsequence_region = mutated_sequence[start_index:start_index + len(subsequence)]


# Print result
if 'M' in subsequence_region:
    print("After 1000 reproduction cycles, the subsequence has been mutated.")
else:
    print("After 1000 reproduction cycles, the subsequence is still present.")
```

2 d. Using a for loop, repeat this simulation 100 times, recording whether DNAsubsequence has been mutated each time. Use this to calculate the probability of DNAsubsequence being mutated. 15

```python
import random

# Read main sequence file
file1 = open('DNAsequence_formative.txt', 'r')
main_sequence = file1.read().strip()
file1.close()

# Read subsequence file
file2 = open('DNAsubsequence_formative.txt', 'r')
subsequence = file2.read().strip()
file2.close()


# Find the position of subsequence in main sequence
start_index = main_sequence.find(subsequence)
subsequence_length = len(subsequence)

# Counter for number of times subsequence is mutated
mutated_count = 0

# Repeat simulation 100 times
for simulation in range(100):
    # Convert main sequence to list for mutation simulation
    dna_list = []
    for base in main_sequence:
        dna_list.append(base)
    
    # Simulate 1000 reproduction cycles
    cycles = 1000
    mutation_rate = 1/10000
    
    for cycle in range(cycles):
        i = 0
        while i < len(dna_list):
            # Only check bases that haven't mutated yet
            if dna_list[i] != 'M':
                # Generate random number and check if mutation occurs
                random_number = random.random()
                if random_number < mutation_rate:
                    dna_list[i] = 'M'  # Mark as mutated
            i += 1
    
    # Convert mutated sequence back to string
    mutated_sequence = ''
    for i in dna_list:
        mutated_sequence += i
    
    # Extract the subsequence region from mutated sequence
    subsequence_region = mutated_sequence[start_index:start_index + subsequence_length]
    
    # Count if mutated
    if 'M' in subsequence_region:
        mutated_count += 1

# Calculate probability of mutation
mutation_probability = mutated_count / 100

# Print results
print("The probability of the subsequence being mutated after 1000 reproduction cycles is",mutation_probability)
```
