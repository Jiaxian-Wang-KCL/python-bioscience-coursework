```python
import random

def guess_number_game():
    number_to_guess = random.randint(1, 10)
    guess = None
    attempts = 0
    max_attempts = 3
    score = 0
    print("Welcome to the guess the number game! You have 3 attempts to guess.Please pick a number between 1 and 10.")
    
    for attempt in range(1, max_attempts + 1):
        try:
            guess = int(input(f"This is your {attempt} Guess，Enter your guess： "))
            if guess < number_to_guess:
                print("Too low!")
            elif guess > number_to_guess:
                print("Too high！")
            else:
                # 根据第几次猜中来加分并提供不同的祝贺信息
                if attempt == 1:
                    score = 3
                    print(f"Congratulations, you guessed correctly on the first try! You have won the {score} scores！")
                elif attempt == 2:
                    score = 2
                    print(f"Congratulations, you guessed correctly on the second try! You have won the {score} scores！")
                else:
                    score = 1
                    print(f"Well done! The third time you finally guessed correctly! You get {score} scores")
                break
        except ValueError:
            print("Please enter a valid number！")

    # 如果三次都没猜对
    if score == 0:
        print(f"I'm sorry you used up all your chances. The correct number is {number_to_guess}.")

# 运行游戏
guess_number_game()

    



```
