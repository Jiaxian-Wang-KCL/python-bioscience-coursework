```python
#This is a program to allow the user to list their top 10 songs
top10songlist = []

for i in range(0,10):
    if i <= 6 :
        top10songlist.append(input("List your "+str(10-i)+"th favourite song: "))
    elif i ==8:
        top10songlist.append(input("List your "+str(10-i)+"nd favourite song: "))
    elif i == 9:
        top10songlist.append(input("List your "+str(10-i)+"st favourite song: "))
    else:
        top10songlist.append(input("List your "+str(10-i)+"rd favourite song: "))



print("This is the list of your top 10 songs, ",top10songlist)
```
