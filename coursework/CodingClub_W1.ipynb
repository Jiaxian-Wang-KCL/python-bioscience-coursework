```python
user_temp = int(input("Enter your temperature:"))
if user_temp < 0:
    print("Freezing")
elif 0 <= user_temp <= 20:
    print("Cold")
elif 20< user_temp <= 30:
    print("Warm")
else:
    print("Hot")
```

## Working with Files

1. Randomly generate a DNA sequence 100 nt long and store in variable `seq`.
2. Write the DNA sequence to a file named **seq.txt**.
3. Read the DNA sequence in from the file and store in variable `read_seq`.

```python
import random

DNA_nucleotide = ["A", "T", "C", "G"]
seq = ''
for i in range(100):
    seq += random.choice(DNA_nucleotide)
print(seq)
file = open("seq.txt","w")
file.write(seq)
file.close()
file1 = open("seq.txt","r")
read_seq = file1.read().strip()
print(read_seq)
```
