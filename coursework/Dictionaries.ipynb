# Dictionaries - what are they and how do they work?


## What are dictionaries?

### 1. Dictionaries - like lists, but not just numbers

Remember lists? Lists contain a list of things which can be indexed using a number, that is, you access an item in the list by using the corresponding number. Here's an example of using a list:

```python
import random

#here I am setting up my list of various items and calculating its length
this_is_a_list = ["apple","pear","banana","bike tyre"]
lenlist = len(this_is_a_list)

#set up Boolean flag to control while loop
#note that yuk the Boolean variable is different to yuk the string
yuk = False

#while the user has not inputted yuk, present random items from the list.
while yuk==False:
    random_list_index = random.randint(0,lenlist -1)
    test_eat = input("Would you eat "+ this_is_a_list[random_list_index]+". Input yum or yuk. ")
    if test_eat=="yuk":
        yuk = True
    elif test_eat=="yum":
        print("Excellent!")
    else:
        test_eat = input("Please enter another response.")

print("We found something that you wouldn't eat.")
```

Dictionaries are like lists, except instead of just being able to index using a number, you can index using any immutable variable. This could be a number, or a string, or even a tuple. (But not a list! Because lists are mutable). When making a list, it's obvious what the indices are, you start at 0 and count up, so you don't need to tell Python. With dictionaries, there is no obvious starting point or notion of next, so you have to tell python what the indices are. 

Dictionaries are set up using curly brackets, rather than square brackets like a list. Each bit inside the curly brackets consists of two things with a colon between them. The first of those is the thing that will be used as the index. This is called a key. The second is the thing that is being referred to. This is called the value. This type of structure can also be referred to as an associative array, as it associates two things and uses one to call the other.

Below I've rewritten the program above to function with a simple dictionary that functions in the same way as a list.

```python
import random

#create dictionary with keys of numbers and values of strings corresponding to the items from the list
this_is_a_dictionary = {0:"apple",1:"pear",2:"banana",3:"bike tyre"}
lendict = len(this_is_a_dictionary)

#create boolean flag
yuk = False

#while boolean flag not triggered, select a random key from the dictionary and ask if user would eat
#corresponding value. If response is yuk then exit loop
while yuk==False:
    random_dict_key = random.randint(0,lendict -1)
    test_eat = input("Would you eat "+str(this_is_a_dictionary[random_dict_key])+". Input yum or yuk. ")
    if test_eat=="yuk":
        yuk = True
    elif test_eat=="yum":
        print("Excellent!")
    else:
        test_eat = input("Please enter another response.")

print("We found something that you wouldn't eat.")
```

The example shows that there is a lot in common between lists and dicts. There is so much in common that in some other languages all lists are just dicts. 

### 2. Dictionary keys and values

The big advantage of dictionaries is that you can index dictionary using things which are not numbers. Below I create a dictionary where the key is the item being considered as to whether it would be pleasant to eat or not. You can print out both the keys and the values present within the dictionary.

```python
eating_dict = {"apple":"yum","pear":"yum","banana":"yum","bike tyre":"yuk"}

#print out the dictionary keys
print(eating_dict.keys())
print(eating_dict.values())

#call from the dictionary using a key
print(eating_dict["apple"])
print(eating_dict["bike tyre"])
```

You can use the keys to obtain the value corresponding to the key within a program. Here's an example:

```python
eating_dict = {"apple":"yum","pear":"yum","banana":"yum","bike tyre":"yuk"}

exitflag = False

#while the supplied key is in the dictionary, return the value. Otherwise print a message indicating the item was absent
while exitflag==False:
    input_item = input("Enter the item that you want to know whether it's good to eat: ")
    if input_item in eating_dict:
        print(eating_dict[input_item])
    else:
        print("Sorry, that item isn't in my dictionary.")
        exitflag=True
        
```

### 3. Dictionaries are mutable (just like lists)

```python
#set up the dictionary and measure it's length
this_is_a_dictionary = {"apple":"yum","pear":"yum","banana":"yum","bike tyre":"yuk"}

#set up flag for loop
continue_loop = True

#set up loop to add new key/value pairs to dictionary. Terminate if value not permitted.
while continue_loop:
    newkey = input("What foodstuff or item would you like to enter in the database?")
    newvalue = input("Is this item yum or yuk?")

    this_is_a_dictionary[newkey] = newvalue
    if newvalue !="yum" and newvalue !="yuk":
        continue_loop = False
        

print("You didn't return an appropriate response.")
```

### 4. Dictionaries and duplication

A key can only appear once in a dictionary. If you put a key into a dictionary twice, two things can happen:

```python
#if you have two definitions with the same key and value, they are equivalent and will only appear once
#if you print out the dictionary
this_is_a_dictionary = {"apple":"yum","apple":"yum","pear":"yum","banana":"yum","bike tyre":"yuk"}
print(this_is_a_dictionary)
#if you have two definitions with the same key and different value, only the last value will be used.
this_is_another_dictionary = {"apple":"yum","apple":"yuk","pear":"yum","banana":"yum","bike tyre":"yuk"}
print(this_is_another_dictionary)
```

### 6. Loops

`eating_dict.keys()` returns all of the keys as a list, so you can iterate over the list and index into the dict:

```python
eating_dict = {"apple":"yum","pear":"yum","banana":"yum","bike tyre":"yuk"}

#print out the dictionary keys
for k in eating_dict.keys():
    print("The " + k + " is " + eating_dict[k])
```

This is the most common and most useful way of iterating through dicts, so Python has a shorthand, you don't need to write `.keys()`:

```python
for key in eating_dict:
    print("The " + key + " is " + eating_dict[key])
```
