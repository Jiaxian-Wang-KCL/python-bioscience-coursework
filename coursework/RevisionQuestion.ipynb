## Restriction enzyme finder

Restriction enzymes bind to specific DNA sequences and cut the DNA strand. The table below shows the recognition sequences for some common restriction enzymes with the cut site indicated by the slash (/). For example, BamHI recognises the sequence `GGATCC` and cuts between the first two nucleotides of that sequence.

**Table 1: Restriction Enzymes Recognition Sequences**

| Sequence | Enzyme|
|:---------|-------|
| G/GATCC  | BamHI |
| G/AATTC  | EcoRI |
| GAT/ATC  | EcoRV |
| G/CTAGC  | NheI  |
| T/CTAGA  | XbaI  |

3. Write a function to identify the cuts sites for any of the enzymes in table 1. The user should provide the function with a DNA sequence and the name of the enzyme. The function should return a list containing the position of the cut sites.

````python
file = open("seq1(1).txt", "r")
seq = file.read().strip()
file.close()
def find_cut_site(sequence, enzyme_name):
    enzyme_dict = {	"BamHI":"G/GATCC",
	"EcoRI":"G/AATTC",
	"EcoRV":"GAT/ATC",
	"NheI":"G/CTAGC",
	"XbaI":"T/CTAGA"}
    enzyme_seq = enzyme_dict[enzyme_name]
    re_enzyme_seq = ''
    enzyme_cut_sites = []
    for i in range(len(enzyme_seq)):
        if enzyme_seq[i] == "/":
            cut_pos = i - 1
        else:
            re_enzyme_seq += enzyme_seq[i]
    for i in range(len(sequence) - len(re_enzyme_seq) + 1):
        if sequence[i:i + len(re_enzyme_seq)] == re_enzyme_seq:
            cut_site = i + cut_pos
            enzyme_cut_sites.append(cut_site)
    print(enzyme_name)
    print(enzyme_cut_sites)

find_cut_site(seq,"BamHI")    
find_cut_site(seq,"EcoRI")
find_cut_site(seq,"EcoRV")
find_cut_site(seq,"NheI")
find_cut_site(seq,"XbaI")
````

#### Restriction enzymes with amiguity codes

Some restriction enzymes have ambiguous recognition sequences where more than one type of nucleotide can be accommodated at certain positions in the recognition sequence.

**Table 2: Restriction Enzymes Recognition Sequences**

| Sequence     | Enzyme|
|:-------------|-------|
| G/GYRCC      | BanI  |
| GCCNNNN/NGGC | BglI  |
| W/CCGGW      | BsaWI |

**Table 3: IUPAC Ambiguity Codes for Nucleotides**

| Code | Bases      |
|:----:|------------|
| A    | A          |
| C    | C          |
| G    | G          |
| T    | T          |
| W    | A or T     |
| S    | C or G     |
| R    | A or G     |
| Y    | C or T     |
| N    | A C G or T |

4. Write a function to identify the cut sites for restriction enzymes with ambiguous recognition sequences. Your function should take the recognition sequence exactly as it is written in Table 2 and the name of a sequence file to search. For each cut site found, your function should print the position of the cut site, and the 5 nucleotides either side of each cut site with a space representing where the DNA strand has been cut.

````python
file = open("seq1(1).txt", "r")
seq = file.read().strip()
file.close()
def find_ambi_cut_site(sequence, enzyme_name):
    enzyme_dict = {"BanI":"G/GYRCC", "BglI":"GCCNNNN/NGGC", "BsaWI":"W/CCGGW"}
    code_dict = {"A":"A","C":"C","G":"G","T":"T","W":["A", "T"],"S":["C", "G"],"R":["A", "G"],"Y":["C", "T"],"N":["A","C", "G", "T"]}
    enzyme_seq = enzyme_dict[enzyme_name]
    re_enzyme_seq = ''
    enzyme_cut_sites = []
    for i in range(len(enzyme_seq)):
        if enzyme_seq[i] == "/":
            cut_pos = i - 1
        else:
            re_enzyme_seq += enzyme_seq[i]
            
    for i in range(len(sequence) - len(re_enzyme_seq) + 1):
        match_point = 0 
        for j in range(len(re_enzyme_seq)):
            ambi_code = re_enzyme_seq[j]
            ambi_list = code_dict[ambi_code]
            if sequence[i + j] in ambi_list:
                match_point += 1
        
            
        if match_point == len(re_enzyme_seq):
            cut_site = i + cut_pos
            enzyme_cut_sites.append(cut_site)
    print(enzyme_name)
    print(enzyme_cut_sites)
    print(sequence[cut_site-4:cut_site+1],sequence[cut_site+1:cut_site+6])
find_ambi_cut_site(seq, "BanI")
find_ambi_cut_site(seq, "BglI")
find_ambi_cut_site(seq, "BsaWI")
````
