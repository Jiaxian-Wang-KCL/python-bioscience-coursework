#### Background

CRISPR is a recent technology which exploits a bacterial anti-viral system to produce a programmable nuclease. This allows experimenters to cut genomic DNA at precise locations, making a whole range of genetic manipulations much simpler. The major advance of CRISPR over earlier technologies is that the cutting site is determined by an RNA molecule (small guide RNA or sgRNA) which can be easily synthesised.

The diagram below shows the key features of how Cas9 binding to double-stranded DNA is targeted by the the sgRNA. The forward and reverse strand are the two strands of the target DNA which have separated due to Cas9 binding. The sgRNA anneals to the part of the reverse strand which it is complementary to while the Cas9 protein binds to the adjacent PAM. The PAM (protospace adjacent motif) is a 3 nucleotide motif. The PAM sequence for SpCas9, the most commonly used Cas9, is NGG, meaning the first nucleotide can be A,G,C,or T, but the following two nucleotides must be G. Once bound, the Cas9 protein cuts both strands of DNA at the sites indicated.

```
                            cut  PAM
                             \/  ┌─┐
5` - ...GCCTAACGTGATGCCGACGTTCGACAGGTAGCC... - 3' forward strand

5' -         ACGUGAUGCCGACGUUCGAC...         - 3' sgRNA
3' - ...CGGATTGCACTACGGCTGCAAGCTGTCCATCGG... - 5' reverse strand
                             /\
                             cut
        |    |    |    |    |    |    | 
```

The PAM sequence is bound by the Cas9 protein, so is not included in the sgRNA. The sgRNA sequence must correspond to the 20 nucleotides upstream of (before) the PAM motif for the Cas9 to be able to bind to the DNA, as shown in this diagram.

DNA sequence files only contain the forward strand. To design gRNA sequences, you first find an NGG motif in the sequence, then take the 20 nucleotides upstream of it like so. It is simpler to express the gRNA sequence as a DNA sequence (with T rather than U) because we often produce the sequence as DNA oligonucleotides which are later transcribed into RNA.

```
GCCTAACGTGATGCCGACGTTCGACAGGTAGCC - forward strand
     └──────────────────┘└─┘
        20 nucleotides   PAM
     ┌──────────────────┐
     ACGTGATGCCGACGTTCGAC         - gRNA sequence as DNA
```

````python
pam_sequences = []

# Loop through the sequence looking for GG patterns
for i in range(len(dnaseq) - 2):  # -2 to ensure we don't go out of range
    # Check if current position and next position form "GG"
    if dnaseq[i+1] == 'g' and dnaseq[i+2] == 'g':
        # Get the full PAM sequence (NGG)
        pam_sequence = dnaseq[i] + dnaseq[i+1] + dnaseq[i+2]
        pam_sequences.append((i, pam_sequence))
if pam_sequences:
    print(f"Found {len(pam_sequences)} PAM sequence(s):")
    print("Index\tPAM\tSequence Context")
    print("-" * 40)
    
    for index, pam in pam_sequences:
        # Show some context around the PAM
        start = max(0, index - 5)
        end = min(len(dnaseq), index + 8)
        context = dnaseq[start:end]
        
        # Mark the PAM in the context
        pam_start_in_context = index - start
        marked_context = (context[:pam_start_in_context] + 
                         "[" + context[pam_start_in_context:pam_start_in_context+3] + "]" + 
                         context[pam_start_in_context+3:])

        print(f"{index}\t{pam}\t{marked_context}")
else:
  print("No PAM sequences (NGG pattern) found in the DNA sequence.")
````
