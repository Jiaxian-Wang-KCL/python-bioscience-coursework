8. A second DNA sequence has been provided (dna2.txt), which contains a single point mutation (substitution) compared to dna1.txt. Read this sequence in and compare it nucleotide by nucleotide to the first sequence to find the location of the mutation. Print the position of the point mutation and both the original and mutated nucleotide. (10)

```python
# Read both DNA sequences
try:
    with open("dna1.txt", "r") as file:
        dna1 = file.read().strip().upper()
    with open("dna2.txt", "r") as file:
        dna2 = file.read().strip().upper()
except:
    print("Error reading files")
    exit()

# Check sequence lengths
if len(dna1) != len(dna2):
    print("Sequences have different lengths!")
    exit()

print("Comparing dna1.txt and dna2.txt...")
print()

# Find the point mutation
i = 0
while i < len(dna1):
    if dna1[i] != dna2[i]:
        print(f"Point mutation found at position {i+1}")
        print(f"Original (dna1.txt): {dna1[i]}")
        print(f"Mutated  (dna2.txt): {dna2[i]}")
        break
    i += 1
else:
    print("No mutations found - sequences are identical")
```
