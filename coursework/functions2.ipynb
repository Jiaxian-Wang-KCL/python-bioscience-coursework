## Writing your own functions

You've already encountered functions, things like `print`, `len` and so on. Each time you call `print`, Python runs a big block of code which has already been written for you in order to display things to the screen.

You can also deifne your own functions. These allow you to package up some code and re-use it repeatedly. This saves on a lot of copy and pasting and makes big programs easier to understand becaue you can give each block of code a meaningful name.

### Defining a function

You define functions with `def`. The syntax is quite similar to loops and ifs, there's a `:` at the end of the line and everything belonging to the function is indented.

```python
#Define a function
def display_big_message():
    print("┌──────────────────────────┐")
    print("│ Hello! This is a message │")
    print("└──────────────────────────┘")
    
#Call the function several times
display_big_message()
display_big_message()
display_big_message()
display_big_message()
```

### Function arguments

`display_big_message()` is fine but it can do just one thing so it's not especially useful. Functions are more useful when they have arguments that you pass to them, like `print`:

```python
#define a function with one argument
def message_box(message):
    #Make a line of -'s of the right length
    dashes = "─" * len(message)
    print("┌" + dashes + "┐")
    print("│" + message +"│")
    print("└" + dashes + "┘")
    
message_box("Hello")
message_box(" Alert!!! ")
```

You can have as many arguments as you like.

### Return values

Functions are even more useful when you can compute something and return it. You have already encountered this with `len`, it computes the length of a string, list or dict and returns the value, which you can then use. To return a value, you use `return`:

```python
#Define a fnuction to reverse a string
def reverse_string(message):
    backwards = ""
    for i in range(len(message)):
        backwards += message[len(message)-1-i]
    return backwards
    
print("This text is " + reverse_string("back to front"))    
print(reverse_string("a man a plan a canal panama"))    
print(reverse_string("eva can i stab bats in a cave"))    
print(reverse_string("was it a rat i saw"))    
```

### Function chaining

If you have a two functions, `func1` and `func2` each of which take in a value and each of which return a value, you can call one after the other by chaining functions. You don't need to assign the result from `func1` to a variable, you can pass it straight to func2.
You would right this as `result = func2(func1(input))`. Here is an example:

```python
#Define a function to replace letters in a string with letters
#that look mirrored
def mirror_letters(text):    
    backwards_letters = {'a': 'ɒ', 'b': 'd', 'c': 'ɔ', 'd': 'b', 'e': 'ɘ', 'f': 'ʇ', 'g': 'ϱ', 'h': 'ʜ', 'i': 'i', 'j': 'į', 'k': 'ʞ', 'l': '│', 'm': 'm', 'n': 'n', 'o': 'o', 'p': 'q', 'q': 'p', 'r': 'ɿ', 's': 'ƨ', 't': 'Ɉ', 'u': 'υ', 'v': 'v', 'w': 'w', 'x': 'x', 'y': 'γ', 'z': 'z'}
    result = ""
    for i in text:
        if i in backwards_letters:
            result += backwards_letters[i]
        else:
            result += i
    return result

print(mirror_letters("look at this in a mirror"))
print(mirror_letters(reverse_string("look at this in a mirror")))
```

### Local variables and scope

This is a very important topic and programs can behave in a confusing way if you do not understand it. Variables defined in the main program outside of functions can be used anywhere and so these are called "global". They exist in the global scope.

I'm going to explain scopes using boxes and labels as in the video lectures. But now the labels have colours and the colours are significant. For example:

```python
# Switch to using blue post it notes inside the function
def times_3_plus_one(x):
    y = x * 3 # Put (x+1) in a box and stick a BLUE post it note on labelled "b"
    return y+1 
    #Find all the BLUE labels and remove them.

a = 1     # Put 1 in a box and stick a RED post-it note on labelled "a"

b = times_3_plus_one(a)
print(b)
```

Running through the program:
* Define a function called times_3_plus_1
* Put 1 in a box and label it "a" with a red note
* Call the function with "a" as an argument.
* Switch to using blue notes because you're now in a function
* The first argument of the function is "x" so stick a blue "x" on box "a"
* Compute 1*3 (it's 3) and put 3 in a box labelled with a blue y
* compute 3+1 (it's 4) and stick it in a box, which is to be returned
* The function returns so remove all the blue notes
* Take the returned box and label it with a red "b"
* print b

So what happened to `y`? Well `y` was defined inside a function so it had a blue label. When you leave the function you remove all the blue notes and throw them away. So if you now try to use `y`, you get an error:

```python
print(y)
```

`y` is called a "local variable" because it's local to the function, is not present outside the function.

### Shadowing

Shadowing is when a local variable has the same name as a global one. Functions will always prefer local variables when there's a choice. For example:

```python
x = 1
message = "That's the result"

def add_2_and_print(x):
    print(x+2)
    print(message)
    
a=3
add_2_and_print(a)
```

You can figure out what's going on by running through the program using boxes:
* Put 1 in a box and label it with a red note with `x` written on
* put "That's the result" in a box and label it with a red note with `message` written on
* define a function called `add_2_and_print`
* Put 3 in a box and label it with a red a
* Call `add_2_and_print`, so switch to using blue labels
* We're using <span style="color:red">a</span> as the argument, the argument is called `x` so put a blue x on box as well
* Look for a box with a blue `x` on it, it has a 3 inside. Add two and print.
* Look for a box with a blue `message` on it. There isn't one, so look for a box with a red `message` on it.
* print the contents
