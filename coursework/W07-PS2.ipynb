We're going to use the example of the Fibonacci sequence to revise the pre-session material on creating functions in python.

The Fibonacci sequence goes as follows:

$$0, 1, 1, 2, 3, 5, 8, 13, ...$$

Fibonacci numbers can be defined mathematically as $F_n$ where $n$ is the position in the sequence and $F$ is the value at that position. The sequence starts with:

$$F_0 = 0$$
$$F_1 = 1$$

and then continues according to the formula:
$$F_n = F_{n-2} + F_{n-1}$$

#### 1. Simple Fibonacci Sequence

For the first step, create a simple program using a **for loop** to generate the first 10 values in the Fibonacci sequence (ie. $F_0$ to $F_9$).

- Start by creating a list containing the first 2 values: 0 and 1 (these are the values of positions 0 and 1 in the sequence)
- Create a **for loop** which:
    - runs from positions 2 to 9 of the sequence
    - appends the Fibonacci number for that position to the end of the list
- Then print the whole list of values

*Tip: Fibonacci numbers and python lists both start numbering from zero. So in a python list containing the Fibonacci sequence,* `list[0]` *will contain* $F_0$ *and* `list[9]` *will contain* $F_9$

```python
# start sequence
F0 = 0
F1 = 1
Fibo_list = [F0,F1]
# loop through generating sequence
for i in range(8):
    Fibo_num = Fibo_list[i]+Fibo_list[i+1]
    Fibo_list.append(Fibo_num)
# print sequence
print(Fibo_list)
```

#### 5. Fibonnacci Function which returns a value

Now, rather than printing the output, get your function to return the value. Also, rather than returning the whole sequence, get it to return just the nth value.

So if you call `fib(6)` the function will return the value of $F_6$ which is `8`

```python
# define function
def fib(num):
    # start sequence
    F0 = 0
    F1 = 1
    Fibo_list = [F0,F1]
    # loop through generating sequence
    for i in range(num-1):
        Fibo_num = Fibo_list[i]+Fibo_list[i+1]
        Fibo_list.append(Fibo_num)
    # print sequence
    return Fibo_list[num]
# call function
print(fib(6))
```

#### 7. How many pairs of rabbits?

The Fibonacci sequence was originally described (at least in European history) as the number of pairs of rabbits produced from an idealised breeding experiment.
- One pair of new-born rabbits is placed in a field at month 1 = 1 pair present
- At month 2, they are one month old and breed = still 1 pair present
- At month 3, the pregnant rabbit gives birth to a pair of rabbits = 2 pairs present
- At month 4, the first pair give birth to a third pair of rabbits, the second pair mate but do not give birth = 3 pairs present
- etc...

This simplified example ignores several important biological aspects of rabbit reproduction.

Putting those issues aside for the time being, use your `fib()` function to calculate the total number of rabbits (not pairs) at month 12.

#### 7. A Fibonacci spiral

Fibonacci sequences can be seen all over the natural world from the arrangement of flowers to nautilus shells.

If your `fib()` function is working correctly, the code below will generate a Fibonacci spiral.

```python
import matplotlib.pyplot as plt

# set origin of plot
x = [0]
y = [0]

# append the next value in sequence in the correct direction
for i in range(1, 15):
    if i % 4 == 0:
        x.append(x[i-1] + fib(i))
        y.append(y[i-1])
    elif i % 4 == 1:
        x.append(x[i-1])
        y.append(y[i-1] + fib(i))
    elif i % 4 == 2:
        x.append(x[i-1] - fib(i))
        y.append(y[i-1])
    else:
        x.append(x[i-1])
        y.append(y[i-1] - fib(i))

# generate the plot
plt.plot(x,y)
plt.show()
```
