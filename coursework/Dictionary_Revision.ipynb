# Dictionaries

## 1. Basic dictionary use

#### Creating a Dictionary

Create a dictionary that stores the names of five different species of bacteria as keys and their respective genome sizes (in megabases) as values. Use the abbreviated names as the key - ie. E. coli for *Escherichia coli*

| Bacterial Species              | Genome Size (Megabases) |
|--------------------------------|-------------------------|
| Escherichia coli               | 4.6                     |
| Staphylococcus aureus          | 2.8                     |
| Bacillus subtilis              | 4.2                     |
| Lactococcus lactis             | 2.5                     |
| Mycobacterium tuberculosis     | 4.4                     |s	4.4= 1000 kilobases).

#### Accessing Values

Using your dictionary, write a Python script to print the genome size of *S. aureus*.

#### Updating Values

Using your dictionary, write a Python script to update the genome size of B. subtilis to 4.3 megabases.

#### Adding Key-Value Pairs

Write a Python program to add a new key-value pair to the dictionary, where the key is L. lactis and the value is 2.5 megabases.

#### Removing Key-Value Pairs

Write a Python script to remove the key-value pair for E. coli from your dictionary.

#### Checking for Keys

Write a Python program to check if S. aureus is present in the dictionary.

#### Dictionary Methods

Write a Python script to get a list of all the keys in the dictionary.

#### Iterating Through a Dictionary

Write a program to iterate through the dictionary and print the species name and corresponding genome size for each entry.

## 2. Programming Challenge

Write a program to analyse the DNA sequence in file **SLC16A1.txt**. Your program should count the number of each of the four nucleotides in the sequence and store these in a dicionary where the nucleotide letter is the key and the count is the value. Your program should then calculate the precentage of each of the four nucleotides occuring the sequence and store these in a separate dictionary. Finally, your program should calculate the percentage the four nucleotides in the first 100 nucleotides of the sequence and print a message giving the percentage of each nucleotide in the first 100 nucleotides and the full sequence.
