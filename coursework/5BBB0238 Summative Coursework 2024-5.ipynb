## Introduction

A fundamental first step in gene expression is the recognition of the promoter which initiates transcription by an RNA polymerase. Protein-coding genes in eukaryotic cells are transcribed by RNA polymerase II (pol II). Transcription is initiated at a site called the Transcriptional Start Site (TSS), which is located in the middle of the ~100 base pair (bp) long core promoter. The core promoter contains a number of sequence motifs which various proteins bind to, assembling the pre-initiation complex. This includes pol II and allows transcription to start.

In this coursework you will use Python to analyse several promoter sequences.

## Generating synthetic promoters

8. Most of our understanding of how promoter sequences regulate transcription have come from studying existing promoters across a range of species. Another approach is to generate synthetic promoter sequences with different combinations of the known motifs and study how effective they are. With current technologies, libraries of tens of thousands of DNA sequences can be synthesized and screened using functional assays.

    Write a program to generate promoter sequences with the following properties:
    
- sequences should be 500 nucleotides in length
- sequences should contain a promoter consisting of a TATA-box and Inr motif with the appropriate spacing
- the TSS should be positioned at nucleotide 450
- the first four nucleotides of all sequences should be AGGT and the last four nucleotides should be CGAA. This is to allow for cloning the sequences using Golden Gate assembly.
- The region between nucleotides 50 and 400 should contain a random selection of between 1 and 5 PDX1, MAFA and/or NKX6-1 binding sites
- all other nucleotides in the sequence should be a random selection of the four nucleotides (A, C, G and T) selected with equal probability
- the positioning of the transcription factor binding sites and the TATA-box should be random within the constraints described above. ie. the TATA-box must be located between 24 and 31 nucleotides upstream of the TSS but this precise position should vary randomly within these constraints rather than being fixed in all sequences.

The program should generate 100 sequences and write them to a file named **promoters.txt**

All nucleotide positions are given in zero-based numbering.

**Note: you should NOT submit the *promoters.txt* file to Keats. Just submit this notebook containing the code to produce the file.**

<p style='text-align: right;'>(20)</p>

```python
import random
def generate_promoter():
    nucleotides = ["A", "C", "T", "G"]
    out = open("promoters.txt","w")
    Y = ["C", "T"]
    R = ["A", "G"]
    for n in range(100):
        seq = []
        for i in range(500):
            seq.append(random.choice(nucleotides))
        seq[0:4] = list("AGGT")
        seq[496:500] = list("CGAA")
        tfbs = {"PDX1":"TCTAAT", 
        "NKX6-1":"TAATTAA",
        "MAFA":"TGCA"
            }
        TSS = 450

        seq[TSS-1] = random.choice(Y)
        seq[TSS] = random.choice(R)
        offset = random.randint(24,31)
        TATA_pos = TSS - offset
        seq[TATA_pos:TATA_pos+5] = list("TATAA")
        num_tf = random.randint(1,5)
        for j in range(num_tf):
            name_key = random.choice(list(tfbs.keys()))
            motif = tfbs[name_key]
            len_motif = len(motif)
            pos = random.randint(50,400-len_motif)
            seq[pos:pos+len_motif] = list(motif)
        seq_string = ""
        for nt in seq:
            seq_string += nt
        out.write(seq_string+"\n")
    out.close()

generate_promoter()
```
