2. Now we will simulate the growth phase. At each timepoint, a proportion of the bacteria will divide to produce two bacteria. The number of additional bacteria are represented by $\delta_N$. The total number of bacteria at the next timepoint ($N_1$) will be the original number bacteria ($N_0$) plus the number of additional bacteria ($\delta_N$). This can be expressed mathematically as:
$$N_1 = N_0 + \delta_N$$

In our model, the number of additional bacteria is controlled by the growth rate, $\mu$.

The number of additional bacteria ($\delta_N$) is determined using both the original number of bacteria ($N$) and the growth rate ($\mu$) as such:
$$\delta_N = \mu N$$

If $\mu = 1$ then all of the bacteria will divide at each timepoint and the total number of bacteria would double each time.

If $\mu = 0.2$ then a fifth of the bacteria will divide at each timepoint and the total number of bacteria would increase more slowly.

The important point to realise is that although we have a constant proportion of bacteria dividing at each timepoint, the number of additional bacteria will increase as the population increases. This is what produces the exponential growth observed in bacterial cultures.

Now you will need to use these equations as a guide to expand your program to include the growth phase. Start by creating variables for each of the variables in the equation. To make it easier to compare code in the class, name your variables:

Start with a value of 0.2 for `mu`.

Plot the number of bacteria for 200 timepoints (including the 50 in the lag phase) and observe how changing $\mu$ changes the final number of bacteria.

Hint: <font color="white">At each timepoint, first calculate the value for **deltaN**, then calculate the new value for **N**.</font>

```python
import matplotlib.pyplot as plt
mu = 0.2
N = 10
N_list = []
t = []
for i in range(50):
    t.append(i)
    N_list.append(10)
for i in range(50,200):
    deltaN = mu*N
    N = deltaN + N
    t.append(i)
    N_list.append(N)
plt.plot(t,N_list)
plt.show()


```
