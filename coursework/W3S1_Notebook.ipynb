### 4. String manipulation

Now you've read in the DNA sequence as a string from the file, let's find out a bit more about it.

That newline character that is included when you read a file can be a bit annoying. It makes finding the length of the string more complicated and can cause a few other problems if you forget about it. If only there was some way we could strip the newline character off the end of the string...

Well there is!

The `strip()` function removes whitespace (including newline) from either end of a string. The simplest way to use it here, is just to include after the `read()` function like so:
```
filecontents.read().strip()
```
Copy your code from the previous cell and modify it to strip off the newline character and count the length of the sequence correctly. Both approaches should give the output `59`. If not, check where you've made a mistake.

````python
###dnaseq_clean = dnaseq.strip()
###length_without_newline = len(dnaseq_clean)
###print("Length without newline:", length_without_newline)
# Open the file dnaseq.txt in read mode
file = open('dnaseq.txt', 'r')

# Read the contents of the file into variable dnaseq and strip newline characters
dnaseq = file.read().strip()

# Close the file
file.close()

# Check the length of the sequence (now without newline characters)
sequence_length = len(dnaseq)

# Print the length of the DNA sequence
print("The length of the DNA sequence is:", sequence_length)

# Also print the sequence itself to verify
print("The DNA sequence is:", dnaseq)
````

Print out each base (denoted by g,t,c or a) in the sequence that you read in from the file with a number corresponding to its position in the sequence. So if your sequence started gttc... then the output would be

    0 g  
    1 t  
    2 t  
    3 c  
    etc.



**Hints:** We've included a few hints in case you need them. Just highlight the text to reveal the hint.

Hint 1: <font color='white'>If you want to go through a string one character at a time, you probably want to use a **for** loop.</font>

Hint 2: <font color='white'>If you want to print several things at the same time, you can include them inside the **print()** function separated by commas. Eg. print(i, dnaseq[i]) will print both the index value (i) and the corresponding letter from the DNA sequence (dnaseq[i]).</font>

````python
for i in range(0,sequence_length):
    print(i," ",dnaseq[i])
````

What you have read in is the forward strand sequence. Now we're going to use the forward strand sequence to write the reverse strand sequence. In the reverse strand sequence the bases are complemented, so in the end you will want a sequence of the same length, but where the letter of each of the bases is changed according to the following rules:

- a becomes t
- t becomes a
- g becomes c
- c becomes g

You should also include a check to see if there is an unexpected letter in the sequence.

#### 🙋‍♂️ Discussion point 🙋‍

With the rest of your group, draw a flow diagram for this program. Include all the steps your code will need to have and any programming structures (eg. if, else, for, while) you will use.

Now convert your flow diagram into code. Check your code runs correctly by comparing the complemented DNA sequence with original sequence you printed out above. If you print both sequences next to each other, you should be able to check that each of the bases is complementary like so:

    old: acgt
    new: tgca

````python
complemented_base = ""
for base in dnaseq:
    if base == "a":
        complemented_base += "t"
    elif base == "t":
        complemented_base += "a"
    elif base == "c":
        complemented_base += "g"
    elif base == "g":
        complemented_base += "c"
    else:
        # Handle unexpected characters
        complemented_base += '?'
        print(f"Warning: Unexpected base '{base}' found")
# Print original and complemented sequences side by side for verification
print("Original:  ", dnaseq)
print("Complement:", complemented_base)

# Alternative: Print them aligned base by base
print("\nBase-by-base comparison:")
for i in range(len(dnaseq)):
    print(f"{i} {complemented_base[i]}")
 
````
