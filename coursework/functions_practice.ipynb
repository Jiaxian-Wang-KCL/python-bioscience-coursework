# Functions

Let's consider something very simple being put into a function. Hopefully this will help understand how functions work without also having to work out what the algorithm inside is doing!

```python
a = 2
a = a+1
print(a)
```

Now let's consider how we would but this into a function. You have to define the function with def, and then you have to give the function name and the inputs in brackets, and you return the number that you want to be outputted. The returned number is assigned to whatever is on the left of the equals sign when the function is called, like this:

```python
def addone(a):
    a = a+1
    return a
a=2
a = addone(a)
print(a)
```

Note everything inside the function is indented one level! If you don't do this you immediately exit the function. Try changing the code above and see.

Don't forget you should always add a descriptive comment above the function definition so anyone using it knows what it's doing

```python
# define a function to add one to the inputted number
def addone(a):
    a = a+1
    return a
a=2
a = addone(a)
print(a)
```

It's important to note that the variables inside the function are local variables, and the variables outside are called global variables. What I did above was use the same name for local and global variables, which is generally a bad idea because two types of variable definitions are interfering with each other. It would be better to do this:

```python
# define a function to add one to the inputted number
def addone(num):
    num = num+1
    return num
a=2
b = addone(a)
print(b)
```

You can have more than one input and more than one output of a function if you want

```python
# define a function which takes two numbers, first output is numbers added and second is numbers multiplied
def addandmult(num1,num2):
    numplus =num1+num2
    nummult = num1*num2
    return numplus,nummult
a=2
b=3
c,d = addandmult(a,b)
print(c,d)
```
