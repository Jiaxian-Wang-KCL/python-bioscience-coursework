2. Generate two lists of numbers as follows:
- x: going from 0 to 9 in increments of 1
- y: going from 10 to 100 in increments of 10

Plot the two lists using pyplot.

```python
import matplotlib.pyplot as plt
x = []
y = []
for i in range(0,10,1):
    x.append(i)
for i in range(10,101,10):
    y.append(i)
print(x)
print(y)
plt.plot(x,y)
plt.show()
```

3. Generate a list starting with 1 and containing 10 values, where each subsequent value is double the previous value.

*Hint: you can either store the current value in a separate variable or lookup the previous value in the list.*

Print the following values:
- whole list
- first item
- last item
- third item
- item in position 3

```python
list = []
for i in range (10):
    list.append(2**i)
print(list)
print(list[0])
print(list[len(list)-1])
print(list[2])
print(list[3])
```
