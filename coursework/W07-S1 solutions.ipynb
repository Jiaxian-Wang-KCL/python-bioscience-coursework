Each UK railway station has a unique three letter code. Here are the codes for some london stations:

| code | station         |
| :--  | :-:              |

| KGX  | Kings Cross     |
| VIC  | Victoria         |

1. Create a dictionary of the station codes and names.

```python
### SOLUTION ###

# create dictionary
stations = {
    'WAT': 'Waterloo',
    'KGX': 'Kings Cross',
    'VIC': 'Victoria',
    'VXH': 'Vauxhall',
    'CLJ': 'Clapham Junction'
}
```

2. Write code to ask the user to enter a code, then print the corresponding station name.

```python
### SOLUTION ###

# ask for input
code = input('Enter a 3-letter station code:')

# find name
name = stations[code]

# return station name
print(name)
```

5. Now write a function that:

- takes a three leter code as an argument
- returns the station name
- is entirely self-contained

Restart the kernel, then test your function. This is a good way to check it's not relying on any existing variables.

```python
### SOLUTION ###

def station_lookup(code):
    
    # create dictionary
    stations = {
        'WAT': 'Waterloo',
        'KGX': 'Kings Cross',
        'VIC': 'Victoria',
        'VXH': 'Vauxhall',
        'CLJ': 'Clapham Junction'
    }
    
    # find name
    name = stations[code]

    # return station name
    return name

# test function
print(station_lookup('WAT'))
print(station_lookup('KGX'))
```
